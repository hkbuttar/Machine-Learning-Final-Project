```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.metrics import classification_report, ConfusionMatrixDisplay
```

```python
# Base dataset
df = pd.read_csv("../Data/Processed/renttherunway_clean.csv")

# GMM clusters
gmm = pd.read_csv("../Data/Processed/gmm_df.csv")[["user_id", "item_id", "body_cluster", "cluster_confidence"]]

# DBSCAN clusters
dbscan = pd.read_csv("../Data/Processed/dbscan_df.csv")[["user_id", "dbscan_cluster"]]

# K-Means clusters
kmeans = pd.read_csv("../Data/Processed/kmeans_clusters.csv").rename(columns={"cluster": "kmeans_cluster"})

# LDA topic features
lda = pd.read_csv("../Data/Processed/renttherunway_lda_topics.csv")
```

```python
df = df.merge(
    dbscan,
    on="user_id",
    how="left"
)
```

```python
df = df.merge(
    gmm,
    on=["user_id", "item_id"],
    how="left"
)
```

```python
df = df.merge(
    kmeans,
    on=["user_id", "item_id"],
    how="left"
)
```

```python
df = df.merge(
    lda,
    on=["user_id", "item_id", "fit", "fit_label"],
    how="left"
)
```

```python
df["is_outlier"] = (df["dbscan_cluster"] == -1).astype(int)
```

```python
NUMERIC_FEATURES = [
    "height_inches",
    "weight_lbs",
    "bmi",
    "age",
    "bust_band",
    "cup_size_num"
]

CLUSTER_FEATURES = [
    "dbscan_cluster",
    "body_cluster",
    "kmeans_cluster"
]

BINARY_FEATURES = ["is_outlier"]

TOPIC_FEATURES = [c for c in df.columns if c.startswith("topic_")]
```

```python
X = df[
    NUMERIC_FEATURES
    + CLUSTER_FEATURES
    + BINARY_FEATURES
    + TOPIC_FEATURES
]

y = df["fit_label"].astype(int)
```

```python
assert y.notna().all()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        # Numeric + topic features
        (
            "num",
            Pipeline([
                ("imputer", SimpleImputer(strategy="median")),
                ("scaler", StandardScaler())
            ]),
            NUMERIC_FEATURES + TOPIC_FEATURES
        ),

        # Cluster features (categorical)
        (
            "cat",
            Pipeline([
                ("imputer", SimpleImputer(strategy="most_frequent")),
                ("onehot", OneHotEncoder(handle_unknown="ignore"))
            ]),
            CLUSTER_FEATURES
        ),

        # Binary features
        (
            "bin",
            SimpleImputer(strategy="constant", fill_value=0),
            BINARY_FEATURES
        )
    ],
    remainder="drop"
)
```

```python
rf = RandomForestClassifier(
    class_weight="balanced",
    n_jobs=-1,
    random_state=42
)
```

```python
pipeline = Pipeline([
    ("prep", preprocessor),
    ("model", rf)
])
```

```python
param_grid = {
    "model__n_estimators": [50, 100],
    "model__max_depth": [6, 8, 12],
    "model__min_samples_leaf": [100, 200]
}
```

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
grid = GridSearchCV(
    pipeline,
    param_grid,
    scoring="f1_macro",
    cv=cv,
    n_jobs=-1,
    verbose=2
)

grid.fit(X_train, y_train)
```

```python
best_rf = grid.best_estimator_

y_test_pred = best_rf.predict(X_test)

print("Best Random Forest parameters:")
print(grid.best_params_)

print("\nRandom Forest — Test Set Performance")
print(classification_report(
    y_test,
    y_test_pred,
    target_names=["Small", "Fit", "Large"]
))
```

```python
fig, ax = plt.subplots(figsize=(6, 6))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_test_pred,
    display_labels=["Small", "Fit", "Large"],
    ax=ax,
    cmap="Blues",
    colorbar=False
)

plt.title("Random Forest — Confusion Matrix (Test Set)")
plt.tight_layout()
plt.savefig("../figures/random_forest_confusion_matrix.png", dpi=300)
plt.show()
```

```python
rf_proba = best_rf.predict_proba(X_test)

rf_proba_df = pd.DataFrame({
    "user_id": df.loc[X_test.index, "user_id"],
    "item_id": df.loc[X_test.index, "item_id"],
    "true_fit": y_test.values,
    "prob_small": rf_proba[:, 0],
    "prob_fit": rf_proba[:, 1],
    "prob_large": rf_proba[:, 2]
})

rf_proba_df.to_csv(
    "../Data/Processed/random_forest_fit_probabilities.csv",
    index=False
)
```

```python
import joblib

joblib.dump(
    best_rf,
    "../Models/random_forest_fit_classifier.joblib"
)
```

```python
# After fitting
prep = best_rf.named_steps["prep"]

feature_names = []

# Numeric + topic features
feature_names.extend(NUMERIC_FEATURES + TOPIC_FEATURES)

# Cluster features (after one-hot)
cat_encoder = prep.named_transformers_["cat"].named_steps["onehot"]
cat_feature_names = cat_encoder.get_feature_names_out(CLUSTER_FEATURES)
feature_names.extend(cat_feature_names)

# Binary features
feature_names.extend(BINARY_FEATURES)
```

```python
importances = best_rf.named_steps["model"].feature_importances_

importance_df = (
    pd.DataFrame({
        "feature": feature_names,
        "importance": importances
    })
    .sort_values("importance", ascending=False)
)
```

```python
importance_df.head(20).to_csv(
    "../Data/Processed/random_forest_feature_importances.csv",
    index=False
)
```
